## Question 1 — Traditional Saving

According to the author, names like `draft_1.doc` or `final_draft.doc` can be a problem because there can be many copies of the same file. It can be difficult to know what was changed, when it was changed, or which file is the most recent version.

A Version Control System (VCS) solves this problem by keeping a history of the changes. It allows researchers to see old versions, compare changes, and go back to a previous version when needed.

## Question 2 — Collaboration and Transparency

### 1. Lab notebook

Git can be used like a digital lab notebook because it keeps a record of the changes made to a project. Researchers can use commit messages to write what they changed or did. This helps them remember the steps of their work.

### 2. Backup and failsafe

Git can also help protect code and files from being lost. By saving the project in a remote place such as GitHub, researchers have another copy of their work. If something happens to their computer, they can recover their files.

## Question 3 — Power of Branching

Branching in Git allows researchers to make a separate copy of their project to try new ideas without changing the main version. This is useful when testing a new statistical method or simulation algorithm.

For example, a researcher can make a new branch and test the new method there. If it works, the changes can be added to the main project. If it does not work, the branch can be removed and the main project will not be affected.
