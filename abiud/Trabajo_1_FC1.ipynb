## Q1: The Problem with Traditional Saving According to the author, why is the traditional method of saving files (e.g., draft_1.doc, final_draft.doc) problematic for scientific research, and what fundamental solution does a Version Control System (VCS) provide?

The traditional method of saving files using names such as draft_1.doc, draft_2.doc, or final_draft.doc is problematic because it is difficult to keep track of which version is the most recent or correct one. It is also difficult to know what changes were made, when they were made, and who made them.

A Version Control System (VCS) solves this problem by keeping a history of changes to files. It allows researchers to record, compare, and recover different versions of their work. This makes scientific research more organized, traceable, reproducible, and secure.

## Q2: Collaboration and Transparency The article lists several "Use cases for Git in science". Choose two of these use cases (e.g., Lab notebook, Backup and failsafe, Mechanism to solicit feedback) and briefly explain how Git improves that specific aspect of scientific work.

### 1. Lab Notebook

Git can be used as a digital lab notebook because researchers can record changes to their code, analyses, and other research files through commits. Each commit records what was changed and when it was changed. This creates a chronological history of the research process.

### 2. Backup and Failsafe

Git provides an effective backup and failsafe mechanism. A local Git repository keeps a history of the project, while a remote repository such as GitHub provides an additional copy of the files. If a computer is damaged or files are accidentally deleted, previous versions can be recovered from the repository. This reduces the risk of losing important scientific work.


## Q3: The Power of Branching Briefly explain the concept of "branching" in Git. Why is this feature particularly useful for researchers who want to test a new statistical method or a new simulation algorithm?

Branching in Git is a feature that allows researchers to create a separate line of development from the main project. Researchers can modify the code, test new ideas, or experiment with different approaches without changing the stable version of the project.

This is particularly useful when testing a new statistical method or simulation algorithm. A researcher can create a new branch, implement and test the new method, and compare its results with the original version. If the new method works well, the changes can be merged into the main branch. If it does not work, the experimental branch can be discarded without affecting the original project.
