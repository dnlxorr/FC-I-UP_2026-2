### Q1: The Problem with Traditional Saving
**According to the author, why is the traditional method of saving files (e.g., draft_1.doc, final_draft.doc) problematic for scientific research, and what fundamental solution does a Version Control System (VCS) provide?**

The traditional method of saving files is problematic because it leads to cluttered, confusing directories filled with multiple versions of the same file. It relies on human memory to track what changes were made, when, and by whom, which is highly prone to error, accidental data loss, or overwriting. For scientific research, this lack of organized history severely hinders reproducibility.

A Version Control System (VCS) like Git provides a fundamental solution by automatically tracking the entire history of a project. Instead of creating new redundant files for every change, VCS takes a "snapshot" of the project at specific points in time. It maintains a single, clean working directory while allowing researchers to revert to any previous state, compare changes, and precisely trace the evolution of their work.

### Q2: Collaboration and Transparency
**The article lists several "Use cases for Git in science". Choose two of these use cases and briefly explain how Git improves that specific aspect of scientific work.**

**1. Lab notebook:** Git functions as an exact, digital lab notebook for computational research. Every commit acts as a permanent record of what was done, including a timestamp, the author, and a descriptive message. This improves scientific work by providing an unambiguous, chronological log of how data was processed and analyzed, ensuring complete transparency and reproducibility.

**2. Backup and failsafe:** When paired with a remote hosting service like GitHub, Git acts as a robust, distributed backup system. If a researcher's local computer crashes, is lost, or corrupted, the entire project—including its complete history—is safely stored remotely and can be cloned to a new machine instantly. This eliminates the risk of losing valuable research progress.


### Q3: The Power of Branching
**Briefly explain the concept of "branching" in Git. Why is this feature particularly useful for researchers who want to test a new statistical method or a new simulation algorithm?**

Branching in Git allows a user to create an independent, parallel line of development that diverges from the main project repository (often called the `main` or `master` branch).

This feature is highly useful for researchers because it provides a safe, isolated environment for experimentation. If a researcher wants to test a new statistical method or tweak a simulation algorithm, they can do so on a separate branch without risking breaking the stable, working code in the main branch. If the new method fails or produces incorrect results, the branch can simply be discarded. If the experiment is successful, the branch can be seamlessly merged back into the main project.
